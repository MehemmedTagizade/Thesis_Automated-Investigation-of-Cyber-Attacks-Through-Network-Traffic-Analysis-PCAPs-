```python
from math import floor
benign_samples = data[data['Label'] == 'BENIGN']
percentage = 0.1
malicious_samples = data[data['Label'] == 'DDoS'].sample(n=floor((begnin_count*percentage)/(1-percentage)), random_state=42)

# Combine the sampled data
data = pd.concat([benign_samples, malicious_samples])

# Shuffle the data
data = data.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Split features and labels
labels = data['Label'].values
features = data.drop(columns=['Label'])

# Encode labels (0 = BENIGN, 1 = DDoS)
y_encoded = np.where(labels == 'BENIGN', 0, 1)

# Scale features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Separate BENIGN and DDoS
X_benign = features_scaled[y_encoded == 0]
X_ddos = features_scaled[y_encoded == 1]

# ✅ Customizable split ratios
train_ratio = 0.6
cv_ratio = 0.2
test_ratio = 0.2

# BENIGN splits
n_benign = X_benign.shape[0]
n_train_benign = int(n_benign * train_ratio)
n_cv_benign = int(n_benign * cv_ratio)
n_test_benign = n_benign - n_train_benign - n_cv_benign

# DDoS splits
n_ddos = X_ddos.shape[0]
n_cv_ddos = n_ddos // 2
n_test_ddos = n_ddos - n_cv_ddos

# Sample indices
np.random.seed(42)
train_benign_idx = np.random.choice(n_benign, size=n_train_benign, replace=False)
remaining_benign = np.setdiff1d(np.arange(n_benign), train_benign_idx)

cv_benign_idx = np.random.choice(remaining_benign, size=n_cv_benign, replace=False)
test_benign_idx = np.setdiff1d(remaining_benign, cv_benign_idx)

cv_ddos_idx = np.random.choice(n_ddos, size=n_cv_ddos, replace=False)
test_ddos_idx = np.setdiff1d(np.arange(n_ddos), cv_ddos_idx)

# Assemble sets
X_train = X_benign[train_benign_idx]

X_cv = np.vstack((X_benign[cv_benign_idx], X_ddos[cv_ddos_idx]))
y_cv = np.concatenate((np.zeros(len(cv_benign_idx)), np.ones(len(cv_ddos_idx))))

X_test = np.vstack((X_benign[test_benign_idx], X_ddos[test_ddos_idx]))
y_test = np.concatenate((np.zeros(len(test_benign_idx)), np.ones(len(test_ddos_idx))))

# Train Isolation Forest on BENIGN only
model = IsolationForest(contamination=0.03, random_state=42)
model.fit(X_train)

# Cross-validation evaluation
cv_preds = model.predict(X_cv)
cv_preds = np.where(cv_preds == -1, 1, 0)

print("=== Cross-Validation Results ===")
print(classification_report(y_cv, cv_preds))
ConfusionMatrixDisplay.from_predictions(y_cv, cv_preds, display_labels=["BENIGN", "DDoS"], cmap='Blues')
plt.title("Cross-Validation Confusion Matrix")
plt.show()

# Final test evaluation
test_preds = model.predict(X_test)
test_preds = np.where(test_preds == -1, 1, 0)

print("=== Test Set Results ===")
print(classification_report(y_test, test_preds))
ConfusionMatrixDisplay.from_predictions(y_test, test_preds, display_labels=["BENIGN", "DDoS"], cmap='Blues')
plt.title("Test Confusion Matrix")
plt.show()
```
